# Decorators in Python

```python
import time
from typing import Callable
from functools import wraps
from typing import Coroutine
import asyncio
from functools import lru_cache
from contextlib import contextmanager
```

```python
# Decorators example
# Add your decorator code here
def timer_deco(func: Callable):
  def wrapper():
    start = time.time()
    res = func()
    end = time.time()
    print("Function process time: ", end - start)
    return res
  return wrapper
```

```python
@timer_deco
def my_func():
  time.sleep(1.5)
  return 124
```

```python
my_func()
```

# Decorators with params

```python
def timer_deco_with_params(func: Callable):
  def wrapper(*args, **kwargs):
    start = time.time()
    res = func(*args, **kwargs)
    end = time.time()
    print("Function process time: ", end - start)
    return res
  return wrapper
```

```python
@timer_deco_with_params
def my_func_with_params(sleep_time: int):
  time.sleep(sleep_time)
  return 124
```

```python
my_func_with_params(3)
```

```python
def limit_calls(limit: int):
  def wrapper(func: Callable):
    @wraps(func)
    def inner(*args, **kwargs):
      """Inner function from decorator"""
      nonlocal limit

      if limit == 0:
        print("Run out of limit!")
        return

      res = func(*args, **kwargs)
      limit -= 1
      return res

    return inner

  return wrapper
```

```python
@limit_calls(3)
def my_func_with_params_v1(sleep_time: int):
  """my_func_with_params_v1 docs"""
  time.sleep(sleep_time)
  return 124
```

```python
my_func_with_params_v1(3)
```

# Decorators for async function

```python
def decorator_with_async(coroutine: Coroutine):
  async def wrapper(*args, **kwargs):
    res = await coroutine(*args, **kwargs)
    return res
  return wrapper
```

```python
@decorator_with_async
async def my_func_async():
  await asyncio.sleep(1)
  return 1
```

```python
await my_func_async()
```

# Decorator with cache

```python
@lru_cache
def my_long_calc():
  time.sleep(3)
  return 33
```

```python
my_long_calc()
my_long_calc()
my_long_calc()
```

# Conext manager

```python
@contextmanager
def ctx_manager():
  print("hello")
  yield
  print("end")
```

```python
with ctx_manager() as man:
  print("123")
```
